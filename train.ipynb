```python
def remove_special_char(text):
    text = text.lower()
    text = sub(r'\d+', ' NUMBER ', text)
    text = sub(r'\S+@\S+', ' EMAIL ', text)
    text = sub(r'http\S+|www\S+|https\S+', ' URL ', text)
    text = sub(r'Subject:|re:|fwd:|fw:', "", text)
    text = sub(r'(Subject|re|fwd|fw)\s*:\s*',"",text, flags=re.IGNORECASE)
    text = sub('[^a-zA-Z0-9]' ," " , text)

    
    text = text.replace('\r\n', ' ').replace('\n', ' ').replace('\r', ' ')

    return text
    
```

```python
dataset['text'] = dataset['text'].apply(remove_special_char)
dataset
```

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords

nltk.download("stopwords")
```

```python
nltk.download("punkt")
```

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
stop_words = set(stopwords.words('english'))
def tokenized(text):
    tokens = word_tokenize(text.lower())
    tokens = [word for word in tokens if word not in stop_words]
    tokens = [stemmer.stem(word) for word in tokens]
    return ' '.join(tokens)

dataset['tokenized'] = dataset['text'].apply(tokenized)
print(dataset[['text', 'tokenized']].head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report



```

```python
vectorizer = TfidfVectorizer(max_features=3000)
x = vectorizer.fit_transform(dataset['tokenized'])
y = dataset['label_num'].values

print(x.shape)
print(dataset['label'].value_counts())
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,y, test_size=0.2, random_state=42,
    stratify=y
)

print(x_train.shape[0])
print(x_test.shape[0])
```

```python
model = MultinomialNB()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)
print(f"\nAccuracy: {accuracy_score(y_test, y_pred):.4f}")

print(f"\nAccuracy: {accuracy_score(y_test, y_pred):.4f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=['ham', 'spam']))
print("\nConfusion Matrix:")
```
